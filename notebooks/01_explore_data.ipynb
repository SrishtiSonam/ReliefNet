# Data Exploration
Analyzing the synthetic disaster datasets.

```python
import pandas as pd
import matplotlib.pyplot as plt

districts = pd.read_csv('../data/districts.csv')
demand = pd.read_csv('../data/synthetic_demand.csv')

print(districts.head())

plt.figure(figsize=(10, 6))
for d in districts['district_id'].unique()[:5]:
    subset = demand[demand['district_id'] == d]
    plt.plot(subset['date'], subset['demand'], label=f'District {d}')
plt.legend()
plt.title('Demand over Time')
plt.show()
```
